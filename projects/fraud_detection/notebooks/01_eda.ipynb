## 1. Load/ Pull the data from DVC

```python
# Define a fucntion to pull the data from DVC
def dvc_pull_data(relative_dvc_path:str):
    # Navigate to the root folder of the project
    project_root = os.path.abspath(os.path.join(os.getcwd(),".."))
    # Setup the file path
    # file_path = os.path.join(project_root, "data", "fraud_data.csv.dvc")
    file_path = os.path.join(project_root, relative_dvc_path)
    # Run the DVC pull command using the full path
    print("DVC pull is complete!")
    return subprocess.run(["dvc", "pull", file_path])
    
                             
```

```python
dvc_pull_data("data/fraud_data_v02.csv.dvc")
```
